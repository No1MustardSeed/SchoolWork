```python
#c.

apple['ticker'] = 'AAPL'
amazon['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
google['ticker'] = 'GOOG'
netflix['ticker'] = 'NFLX'
```

```python
#f

fanng = pd.read_csv('/content/drive/MyDrive/6070/modules/week5/fanng.csv')
fanng['date'] = pd.to_datetime(fanng['date'])
fanng['volume'] = fanng['volume'].astype(int)
fanng = fanng.sort_values(by=['date', 'ticker'])
```

```python
#g

fanng.nlargest(7, 'volume')
```

```python
#h. Use the melt() method to make the 'date' and 'ticker' key identifiers.  Melt the remaining columns so that there are not separated.

melted_fanng = fanng.melt(
    id_vars=['date', 'ticker'],
    value_vars=['open', 'high', 'low', 'close', 'volume']
    )

melted_fanng.head()
```
