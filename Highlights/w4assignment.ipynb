```python
#2b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
tsu_eq_indo = eqdf[(eqdf['place'].str.contains('Indonesia', na=False)) & (eqdf['tsunami'] == 1) & (eqdf['type'] == 'earthquake')]
eq_indo = eqdf[(eqdf['place'].str.contains('Indonesia', na=False)) & (eqdf['type'] == 'earthquake')]

#using the number of rows which is why I am using len
percentage = (len(tsu_eq_indo) / len(eq_indo)) * 100

print(f"{percentage:.2f}%")
```

```python
#d
eqdf['RoF'] = eqdf.parsed_place.str.contains(r'|'.join([
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica',
    'Guatemala', '^Mexico', 'Japan', 'Philippines', 'Indonesia',
    'New Zealand', 'Antarctic', 'Canada', 'Fiji',
    'Alaska', 'Washington', 'California', 'Russia',
    'Taiwan', 'Tonga', 'Kermadec Islands'
]), na=False)

eqdf[['RoF', 'parsed_place']]
```

```python
#e - the number of earthquakes in the Ring of Fire locations and the number outside of them

eq_in_rof = ((eqdf["type"] == "earthquake") & (eqdf["RoF"] == True)).sum()
print(eq_in_rof)

eq_not_rof = ((eqdf['type'] == "earthquake") & (eqdf['RoF'] == False)).sum()
print(eq_not_rof)
```

```python
#f -  Find the tsunami count along the Ring of Fire.

tsu_in_rof = ((eqdf["tsunami"] == 1) & (eqdf["RoF"] == True)).sum()
print(tsu_in_rof)
```
